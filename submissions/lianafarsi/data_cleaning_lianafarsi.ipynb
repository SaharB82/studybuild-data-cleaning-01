```python
import pandas as pd

df = pd.read_excel("First Dataset.xlsx")

print("First 5 rows of the dataset:")
display(df.head())
```

```python
print("Missing values count per column:")
print(df.isnull().sum())
```

```python
 mean_age = df['age'].mean()
df['age'] = df['age'].fillna(mean_age)

 mean_spending = df['total_spending'].mean()
df['total_spending'] = df['total_spending'].fillna(mean_spending)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
duplicates_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates_count}")
```

```python
df = df.drop_duplicates()
print(f"Total rows after removing duplicates: {len(df)}")
```

```python
df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')
text_columns = ['first_name', 'gender', 'city', 'province', 'membership_tier', 'payment_method', 'device', 'discount_used']
for col in text_columns:
  if col in df.columns:
    df[col] = df[col].astype(str).str.strip()

print("Text columns cleaned and dates standardized successfully!")
```

```python
output_file = 'cleaned_dataset_lianafarsi.xlsx'
df.to_excel(output_file, index=False)
print(f"Dataset successfully cleaned and saved as '{output_file}'!")
```

```python
import zipfile
import os

readme_content = """# Data Cleaning Project - First Dataset

This repository contains the first data cleaning project for processing and standardizing customer data using Python (Pandas).

## Project Overview
The goal of this project is to clean a raw dataset, handle missing data, remove duplicates, fix formatting issues, and prepare a clean, analysis-ready dataset.

## Files Included
- data_cleaning_lianafarsi.ipynb
- cleaned_dataset_lianafarsi.xlsx
- README.md
"""

with open('README.md', 'w', encoding='utf-8') as f:
    f.write(readme_content)

zip_filename = 'lianafarsi-data-cleaning.zip'


with zipfile.ZipFile(zip_filename, 'w') as zipf:
    for file in ['data_cleaning_lianafarsi.ipynb', 'cleaned_dataset_lianafarsi.xlsx', 'README.md']:
        if os.path.exists(file):
            zipf.write(file)
            print(f"Added {file} to zip.")

print(f"Zip file '{zip_filename}' created successfully!")
```
